```python
import pandas as pd
from sqlalchemy import create_engine, text
```

```python

engine = create_engine(
    "postgresql+psycopg2://postgres:2605@localhost:5432/postgres"
)

with engine.connect() as conn:
    conn.execution_options(isolation_level="AUTOCOMMIT").execute(
        text('CREATE DATABASE "OLA"')
    )

print("Database OLA created!")
```

```python
from sqlalchemy import create_engine

engine = create_engine(
    "postgresql+psycopg2://postgres:2605@localhost:5432/OLA"
)

with engine.connect() as conn:
    print("Connected to OLA!")
```

```python
df.to_sql(
    "bookings",
    engine,
    if_exists="replace",
    index=False
)
```
